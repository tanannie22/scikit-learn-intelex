```python
from timeit import default_timer as timer
from sklearn.metrics.pairwise import rbf_kernel as sklearn_rbf_kernel
from sklearnex.metrics.pairwise import rbf_kernel as sklearnex_rbf_kernel
import warnings
import numpy as np

warnings.filterwarnings("ignore")

# Generate larger data for testing
X = np.random.rand(1000, 3)
Y = np.random.rand(1000, 3)

start = timer()
sklearn_result = sklearn_rbf_kernel(X, Y)
sklearn_time = timer() - start

print(f"Sklearn Time Taken: {sklearn_time}")
# print(sklearn_result)

start = timer()
sklearnex_result = sklearnex_rbf_kernel(X, Y)
sklearnex_time = timer() - start

print(f"Sklearnex Time Taken: {sklearnex_time}")
# print(sklearnex_result)
```

```python
# compare the results and time taken
print(f"Results are equal: {sklearn_result.all() == sklearnex_result.all()}")
print(f"Sklearnex faster than sklearn: {sklearn_time > sklearnex_time}")
```
